```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree

# Load dataset
data = pd.read_csv("des.csv")

data['Skill_Level'] = data['Skill_Level'].map({
    'Low':0,
    'Medium':1,
    'High':2
})

data['Purchase'] = data['Purchase'].map({
    'No':0,
    'Yes':1
})

X = data[['Hours_Played','Skill_Level','Age']]
y = data['Purchase']

model = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=10,      
    random_state=0
)

model.fit(X, y)

plt.figure(figsize=(24,14))

plot_tree(model,
          feature_names=['Hours_Played','Skill_Level','Age'],
          class_names=['No','Yes'],
          filled=True,
          rounded=True,
          fontsize=20)

plt.title("Decision Tree")
plt.show()
```
